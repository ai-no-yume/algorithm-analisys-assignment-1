```python
import time
import random

#Ingresar la cantidad de elementos
n = int(input("Ingrese la cantidad de elementos: "))

# Llenar la lista con números aleatorios entre 1 y 10000
data = [random.randint(1, 10000) for _ in range(n)]

# Medir tiempo de inicio
start_time = time.perf_counter()

# Ordenar la lista
# data.sort()

# Medir tiempo final
estimated_time = time.perf_counter() - start_time  # tiempo en segundos

# Mostrar solo los primeros 10 elementos (o menos si n < 10)
limit = min(10, n)
print(f"Primeros {limit} datos:", data[:limit])

# Mostrar tiempo de ejecución en segundos con 4 decimales
print(f"Tiempo de ejecución: {estimated_time:.4f} segundos")


```
